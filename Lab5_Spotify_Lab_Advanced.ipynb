```python
# --- 1. Decorator ---
def playback_logger(func):
    """DECORATOR: Adds logging around playback"""
    def wrapper(self):
        print(" Logging playback...")
        result = func(self)
        print(" Playback complete\n")
        return result
    return wrapper

# --- 2. Class 1: Song (Encapsulation + Decorators) ---
class Song:
    total_songs = 0

    def __init__(self, song_id, title, artist, duration):
        self.__song_id = song_id
        self.__title = title
        self.__artist = artist
        self.__duration = duration
        self.__play_count = 0
        Song.total_songs += 1

    def get_title(self):
        return self.__title

    def get_artist(self):
        return self.__artist

    def __increment_plays(self):
        self.__play_count += 1

    @playback_logger
    def play(self):
        self.__increment_plays()
        print(f"Playing: {self.__title} ({self.__play_count})")

# --- 3. Class 2: Playlist (Iterator, Generator, Duck Typing) ---
class Playlist:
    def __init__(self, name):
        self.__name = name
        self.songs = []

    def add_song(self, song):
        self.songs.append(song)

    def __iter__(self):
        """ITERATOR ENTRY POINT"""
        self._index = 0
        return self

    def __next__(self):
        """ITERATOR CONTROL"""
        if self._index >= len(self.songs):
            raise StopIteration
        song = self.songs[self._index]
        self._index += 1
        return song

    def play_songs(self):
        """GENERATOR: Yields songs one at a time"""
        for song in self.songs:
            yield song

    def play_all(self, player):
        """DUCK TYPING: Any object with a play() method works"""
        for song in self:
            player.play(song)

# --- 4. Closures (Stateful Behavior Without Classes) ---
def playback_limiter(max_plays):
    """CLOSURE: Remembers play count across calls"""
    count = 0
    def limiter(song):
        nonlocal count
        if count < max_plays:
            song.play()
            count += 1
        else:
            print(" Playback limit reached")
    return limiter

# --- 5. Duck Typing (Playback Controllers) ---
class SimplePlayer:
    def play(self, song):
        song.play()

class LimitedPlayer:
    def __init__(self, limiter):
        self.limiter = limiter

    def play(self, song):
        self.limiter(song)


# --- 6. Object Creation and Runtime Flow ---
print("--- Creating Songs & Playlist ---")
s1 = Song(101, "Blinding Lights", "The Weeknd", 200)
s2 = Song(102, "Shape of You", "Ed Sheeran", 234)

favorites = Playlist("My Favorites")
favorites.add_song(s1)
favorites.add_song(s2)

print("\n--- Setting up Closure & Players ---")
limit_1 = playback_limiter(1)

normal_player = SimplePlayer()
limited_player = LimitedPlayer(limit_1)

print("\n--- Duck typing in action: Normal Player ---")
favorites.play_all(normal_player)

print("--- Duck typing in action: Limited Player ---")
favorites.play_all(limited_player)
```

Encapsulation & Decorators

A decorator is a function that wraps another function's behavior without changing its core logic. In the code, @playback_logger automatically adds "Logging playback..." before and after a song plays.  

Encapsulation is used to protect the internal state of the Song object.

Iterators, Generators, and Duck TypingAn Iterator enables the for song in playlist loop functionality. It uses __iter__ to set the starting point and __next__ to control the progression and raise StopIteration when done.  A Generator is used to stream songs lazily one at a time using the yield keyword.

Closures (Stateful Behavior Without Classes)A Closure allows an inner function to remember the state of variables in its outer function. In the code, the limiter function remembers the count variable.  Closures demonstrate that you do not always need a class to maintain state. The function itself essentially behaves like an object.

Duck Typing in Action (Controllers)

The Playlist class does not care what type of player is passed to it.  

The only requirement for the player object is that it must have a play(song) method. This allows swapping between a SimplePlayer and a LimitedPlayer seamlessly.

The Iterator is what actively drives the looping behavior.  

The Generator controls the flow of data.
