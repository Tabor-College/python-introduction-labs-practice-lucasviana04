```python
# --- Class 1: Song (Encapsulation) ---
class Song:
    total_songs = 0  # STATIC VARIABLE

    def __init__(self, song_id, title, artist, duration):
        self.__song_id = song_id
        self.__title = title
        self.__artist = artist
        self.__duration = duration
        self.__play_count = 0
        Song.total_songs += 1

    def get_title(self):
        return self.__title

    def get_artist(self):
        return self.__artist

    def set_duration(self, new_duration):
        if new_duration > 0:
            self.__duration = new_duration
            return True
        print("Duration must be positive!")
        return False

    def __increment_plays(self):
        self.__play_count += 1

    def play(self):
        self.__increment_plays()
        print(f"Playing: {self.__title} (Plays: {self.__play_count})")

    @staticmethod
    def get_total_songs():
        return Song.total_songs


# --- Class 2: Playlist (Object Composition) ---
class Playlist:
    def __init__(self, name, creator):
        self.__name = name
        self.__creator = creator
        self.songs = []

    def add_song(self, song):
        self.songs.append(song)
        print(f"{song.get_title()} added to {self.__name}")

    def display(self):
        print(f"\nPlaylist: {self.__name}")
        for i, song in enumerate(self.songs, 1):
            print(f"{i}. {song.get_title()} - {song.get_artist()}")


# --- Class 3: User (Encapsulation + Composition) ---
class User:
    total_users = 0

    def __init__(self, username, email, balance):
        self.__username = username
        self.__email = email
        self.__balance = balance
        self.__playlists = []
        User.total_users += 1

    def pay_for_premium(self, cost):
        if cost <= self.__balance:
            self.__balance -= cost
            print("Premium activated!")
            return True
        print("Insufficient balance!")
        return False

    def create_playlist(self, name):
        playlist = Playlist(name, self.__username)
        self.__playlists.append(playlist)
        return playlist

    @staticmethod
    def get_total_users():
        return User.total_users


# --- Object Creation and Execution Flow ---
print("--- 1. Creating Songs ---")
song1 = Song(101, "Blinding Lights", "The Weeknd", 200)
song2 = Song(102, "Shape of You", "Ed Sheeran", 234)
print(f"Total Songs created: {Song.get_total_songs()}\n")

print("--- 2. Playing Songs ---")
song1.play()
song1.play()
print()

print("--- 3. Creating User & Playlist ---")
user1 = User("Amar", "amar@spotify.com", 500)
favorites = user1.create_playlist("My Favorites")

print("\n--- 4. Adding Songs to Playlist ---")
favorites.add_song(song1)
favorites.add_song(song2)

print("\n--- 5. Displaying Playlist ---")
favorites.display()

print("\n--- 6. Paying for Premium ---")
user1.pay_for_premium(200)
```

Encapsulation (Protecting Data)

How it works: In the Song and User classes, internal data is made PRIVATE by using double underscores (e.g., self.__title, self.__balance).

Why it matters: Private data can never be accessed directly from outside the class. You must use controlled methods (like get_title() or pay_for_premium()) to interact with it. This protects sensitive data and ensures it is validated (e.g., checking if the user has enough balance before activating premium).

Static Members (Class-Level Data)

How it works: Variables like total_songs = 0 and total_users = 0 are declared directly inside the class, outside of the __init__ constructor.

Why it matters: These are static variables. They belong to the class itself, not to individual objects. This allows the system to easily track overarching metrics, like the total number of songs ever created across the entire program.

Object Composition (HAS-A Relationship)

How it works: The Playlist class has an empty list self.songs = []. When add_song() is called, it stores actual Song objects inside this list. Similarly, the User class creates and owns Playlist objects.

Why it matters: This demonstrates how objects interact with and store other objects. A Playlist STORES Songs, and a User OWNS Playlists.

Execution Flow Observations

Constructors: Objects are always brought to life using their __init__ constructors (e.g., Song(101, ...)).

Dot Notation: Methods are called on these objects using dot notation (e.g., song1.play()).

Hidden Logic: The user only calls song1.play(), but internally, the class handles the hidden logic of __increment_plays() to track data accurately without exposing the raw counter to the outside world.
