```python
from abc import ABC, abstractmethod

# --- 1. Abstract Base Class (Vehicle Contract) ---
class Vehicle(ABC):
    @abstractmethod
    def calculate_fare(self, distance):
        pass

    @abstractmethod
    def get_vehicle_type(self):
        pass

class UberX(Vehicle):
    def calculate_fare(self, distance):
        return distance * 10

    def get_vehicle_type(self):
        return "UberX"

uberx = UberX()
print("--- 1. Abstract Base Class ---")
print(f"Fare for 5 miles: {uberx.calculate_fare(5)}\n")


# --- 2. Inheritance (IS-A Relationship) ---
class UberBike(Vehicle):
    def calculate_fare(self, distance):
        return distance * 5

    def get_vehicle_type(self):
        return "UberBike"

bike = UberBike()
print("--- 2. Inheritance ---")
print(f"Is uberx a Vehicle? {isinstance(uberx, Vehicle)}")
print(f"Is bike a Vehicle? {isinstance(bike, Vehicle)}\n")


# --- 3. Runtime Polymorphism ---
print("--- 3. Runtime Polymorphism ---")
vehicles = [UberX(), UberBike()]
for vehicle in vehicles:
    print(vehicle.get_vehicle_type(), vehicle.calculate_fare(10))
print("\n")


# --- 4. Mixin (Electric Feature) ---
class ElectricMixin:
    def charge_battery(self):
        return "Charging battery..."

class UberGreen(ElectricMixin, UberX):
    pass

green = UberGreen()
print("--- 4. Mixin ---")
print(green.get_vehicle_type())
print(green.charge_battery())
print("\n")




class Driver:
    def __init__(self, name, vehicle):
        self.name = name
        self.vehicle = vehicle

    def start_trip(self, distance):
        fare = self.vehicle.calculate_fare(distance)
        return f"{self.name} driving {self.vehicle.get_vehicle_type()} Fare: {fare}"

driver1 = Driver("Alice", UberX())
driver2 = Driver("Bob", UberBike())

print("--- 5. Composition ---")
print(driver1.start_trip(10))
print(driver2.start_trip(10))
print("\n")


# --- 6. Dependency (Uber Service USES Driver) ---
class UberApp:
    def request_ride(self, driver, distance):
        return driver.start_trip(distance)

app = UberApp()
print("--- 6. Dependency ---")
print(app.request_ride(driver1, 5))
```

1. Why can't we instantiate Vehicle?
Because Vehicle is an Abstract Base Class (ABC) that contains abstract methods. Python prevents direct instantiation of ABCs to enforce the rule that a base contract must be fully implemented by its subclasses first.

2. What happens if UberX does not implement get_vehicle_type()?
Python will raise a TypeError at runtime when you try to create an instance of UberX, because it will still be considered an incomplete abstract class.

3. Is ABC enforcing structure at runtime or compile-time?
At runtime. Python checks for the implementation of abstract methods at the exact moment an object of the subclass is instantiated.

4. Why is this useful for large companies like Uber?
It guarantees a strict contract across the codebase. It ensures that any new vehicle type added by developers in the future will definitively have the essential methods (like calculate_fare) needed to integrate with the rest of the application without breaking it.

Inheritance (IS-A Relationship)
   1. Is UberBike automatically a Vehicle?
   Yes. By inheriting from the Vehicle base class, UberBike establishes an "IS-A" relationship, meaning the system recognizes it as a type of Vehicle.


2. What happens if we remove (Vehicle) from UberBike?
UberBike would become a completely standalone class. It would no longer be forced to implement the methods in the contract, and functions like isinstance(bike, Vehicle) would return False.

3. Is inheritance modeling IS-A or HAS-A?
Inheritance models an IS-A relationship.


Runtime Polymorphism (Dynamic Fare Calculation)
1. Where does runtime polymorphism occur?
It occurs inside the for vehicle in vehicles: loop when vehicle.calculate_fare(10) is called. The Python interpreter dynamically figures out which specific method to run (UberX.calculate_fare or UberBike.calculate_fare) depending on the object type at that exact moment.

2. What would happen if method names differed?
The loop would crash with an AttributeError when trying to call a method that doesn't exist on one of the objects. This demonstrates why the ABC contract from Step 1 is so important.

3. Why is this better than using if-else checks?
It makes the code scalable (following the Open/Closed Principle). You can add 50 new vehicle classes to the system, and you will never need to update this loop or write massive if/else blocks to calculate fares.

Mixin (Electric Feature)
1. Why is ElectricMixin listed first?   

In Python's Method Resolution Order (MRO), classes are searched from left to right. Listing the mixin first ensures that if there are any naming conflicts between the mixin and the base class, the mixin's behavior takes priority.

2. Is ElectricMixin modeling IS-A?   

No. Mixins typically model an "IS-CAPABLE-OF" or "CAN-DO" relationship. They append specific behaviors (like charging) without defining the core identity of the object.

3. What happens if both parents define same method?   

Python uses MRO (left to right). The method from the class listed first (in this case, ElectricMixin) will override the method from the second class.

Composition (Driver HAS-A Vehicle)
1. Is Driver inheriting from Vehicle?   

No, Driver is an entirely independent class.

2. What relationship is this modeling?   

It models Composition, which is a HAS-A relationship (a Driver "has a" Vehicle).

3. Why is composition flexible?   

It allows you to assign or swap a vehicle dynamically. For example, Alice can drive an UberX today and switch to an UberBike tomorrow without developers needing to create hardcoded classes like UberXDriver or UberBikeDriver.

Dependency (Uber Service USES Driver)
1. How is dependency different from composition?   

In composition, the object (Vehicle) is stored as an attribute (self.vehicle) and conceptually belongs to the class (Driver). In dependency, the class (UberApp) just receives an object (Driver) temporarily as a parameter to perform a specific task, without storing it.

2. Does UberApp permanently store Driver?   

No, it only uses the driver instance during the brief moment the request_ride method is executing.

3. Why is loose coupling important in real systems?   

It prevents systems from becoming tangled. The UberApp doesn't need to know how a Driver is created or how a Vehicle calculates a fare; it just expects a result. This makes the codebase easier to update, debug, and test.

In two main places: During the simple for vehicle in vehicles iteration, and inside Driver.start_trip(), where the driver calls self.vehicle.calculate_fare() dynamically without caring about what type of vehicle they were assigned.
